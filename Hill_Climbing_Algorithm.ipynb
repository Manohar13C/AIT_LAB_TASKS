```python
import random


def objective_function(x):
  return -(x - 3) ** 2 + 10


def hill_climb(start_x, step_size=0.1, max_steps=500):
  current_x = start_x
  current_val = objective_function(current_x)

  for step in range(max_steps):
    left_x = current_x - step_size
    right_x = current_x + step_size

    left_val = objective_function(left_x)
    right_val = objective_function(right_x)

    if left_val > right_val:
      best_x, best_val = left_x, left_val
    else:
      best_x, best_val = right_x, right_val

    if best_val > current_val:
      current_x, current_val = best_x, best_val
    else:
      break

  return current_x, current_val


start = random.uniform(-5, 10)
peak_x, peak_y = hill_climb(start_x=start, step_size=0.05)

print(f"Starting Point : x = {start:.2f}")
print(f"Discovered Peak: x = {peak_x:.4f}, f(x) = {peak_y:.4f}")
```
