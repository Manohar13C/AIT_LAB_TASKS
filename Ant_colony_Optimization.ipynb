```python
R1 = float(input("Enter pheromone value R1: "))
R2 = float(input("Enter pheromone value R2: "))

total = R1 + R2

P1 = R1 / total
P2 = R2 / total

print("Probability of Path 1 =", P1)
print("Probability of Path 2 =", P2)

if P1 > P2:
    print("Ant prefers Path 1")
elif P2 > P1:
    print("Ant prefers Path 2")
else:
    print("Both paths have equal probability")

print("P1 + P2 =", P1 + P2)
```
